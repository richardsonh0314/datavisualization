The Xenographics project, created by Maarten Lambrechts, is a website that collects weird and unusual charts that most people haven't seen before. Its goal is to create a collection of new and experimental visualizations to inspire people and make new chart types more popular. Instead of just using basic bar charts, line graphs, and pie charts, Lambrechts wants to show that strange looking visualizations can sometimes work better for certain types of data. The project is about what he calls "xenographphobia"  which is when people get scared or confused by charts they've never seen before. The website organizes charts using a star system where "simple weird charts get one star, and very, very weird ones get five stars", so users know how unusual each chart type is compared to normal ones.

What I found most interesting about this project is how it challenges the idea that simple charts are always better. Sometimes a different looking chart that takes more effort to understand can actually show patterns in data that regular charts miss. The project also shows how many new visualization ideas come from mixing different chart types together or borrowing ideas from other areas. My favorite example from the website is the Time Curve (https://xeno.graphics/time-curve/), which takes a normal timeline and bends it into a circle so that similar time periods end up close to each other. This makes it easier to spot repeating patterns over time such as seeing how events that happen every year relate to each other rather than just showing everything in a straight line from past to present.

```python
import warnings
warnings.simplefilter(action='ignore', category=FutureWarning)
import altair as alt
from vega_datasets import data

weather = data.seattle_weather()
weather.head()
```

```python
# Create boxplot
chart = alt.Chart(weather).mark_boxplot(
    size=60,
    opacity=0.8
).encode(
    x=alt.X(
        'weather:N',
        title='Weather Type',
        axis=alt.Axis(labelAngle=45) 
    ),
    y=alt.Y(
        'temp_max:Q',
        title='Maximum Temperature (°C)'
    )
).properties(
    width=400,
    height=300,
    title='Maximum Temperature Distribution by Weather Type'
).resolve_scale(
    color='independent'
)

# Add grid (optional - Altair has subtle gridlines by default)
chart_with_grid = chart.configure_axis(
    grid=True,
    gridOpacity=0.3
).configure_view(
    strokeWidth=0 
)

chart_with_grid
```

##### Question 2 (10 points)

Recreate the figure in Part 3 Question 3 in Homework 1 with Altair. Include necessary title and labels. Here is the question again:
    
    Generate a data visualization to show the relationship between max temperature and min temperature, and then show different color for differnet weather type.

```python
import pandas as pd
import numpy as np

# Create binned data for heatmap using a simpler approach
bin_size = 2
min_temp_min = int(weather['temp_min'].min())
min_temp_max = int(weather['temp_min'].max()) + 2
max_temp_min = int(weather['temp_max'].min())
max_temp_max = int(weather['temp_max'].max()) + 2

# Create bins manually to avoid Interval objects
weather_binned = weather.copy()
weather_binned['temp_min_bin'] = ((weather['temp_min'] - min_temp_min) // bin_size) * bin_size + min_temp_min + bin_size/2
weather_binned['temp_max_bin'] = ((weather['temp_max'] - max_temp_min) // bin_size) * bin_size + max_temp_min + bin_size/2

# Create heatmap data by counting occurrences in each bin
heatmap_data = weather_binned.groupby(['temp_min_bin', 'temp_max_bin']).size().reset_index(name='count')
heatmap_data = heatmap_data.rename(columns={'temp_min_bin': 'temp_min_center', 'temp_max_bin': 'temp_max_center'})

# Create the heatmap layer
heatmap = alt.Chart(heatmap_data).mark_rect(
    opacity=0.3
).encode(
    x=alt.X('temp_min_center:Q', title='Minimum Temperature (°C)'),
    y=alt.Y('temp_max_center:Q', title='Maximum Temperature (°C)'),
    color=alt.Color('count:Q', 
                   scale=alt.Scale(scheme='blues'),
                   legend=alt.Legend(title='Number of Days'))
)

# Create scatter plot layer
scatter = alt.Chart(weather).mark_circle(
    size=60,
    opacity=0.8,
    stroke='white',
    strokeWidth=0.5
).encode(
    x=alt.X('temp_min:Q', title='Minimum Temperature (°C)'),
    y=alt.Y('temp_max:Q', title='Maximum Temperature (°C)'),
    color=alt.Color('weather:N', 
                   title='Weather Type',
                   scale=alt.Scale(range=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd']))
)

# Create diagonal reference line
min_val = weather['temp_min'].min()
max_val = weather['temp_max'].max()

line_data = pd.DataFrame({
    'x': [min_val, max_val],
    'y': [min_val, max_val]
})

diagonal_line = alt.Chart(line_data).mark_line(
    strokeDash=[5, 5],
    stroke='black',
    opacity=0.5,
    strokeWidth=2
).encode(
    x='x:Q',
    y='y:Q'
)

# Combine all layers
combined_chart = (heatmap + scatter + diagonal_line).resolve_scale(
    color='independent'
).properties(
    width=500,
    height=400,
    title='Temperature Relationship Heatmap by Weather Type'
).configure_axis(
    grid=True,
    gridOpacity=0.3
).configure_view(
    strokeWidth=0
)

combined_chart
```

##### Question 3 (10 points)

Create a facet figure in Altair with 2 row and 3 columns. For each panel in the facet figure, it shows a scatter plot between the max and min temperature for one type of weather. Use the wind speed to color the points. Also add interaction to show the max and min temperature and wind speed for each data point. Include necessary title and labels. 

```python
# Create the faceted scatter plot
chart = alt.Chart(weather).mark_circle(
    size=60,
    opacity=0.8,
    stroke='white',
    strokeWidth=0.5
).encode(
    x=alt.X(
        'temp_min:Q',
        title='Minimum Temperature (°C)',
        scale=alt.Scale(nice=True)
    ),
    y=alt.Y(
        'temp_max:Q', 
        title='Maximum Temperature (°C)',
        scale=alt.Scale(nice=True)
    ),
    color=alt.Color(
        'wind:Q',
        title='Wind Speed',
        scale=alt.Scale(scheme='viridis')
    ),
    facet=alt.Facet(
        'weather:N',
        title='Weather Type',
        columns=3,
        header=alt.Header(
            titleFontSize=12,
            labelFontSize=10
        )
    ),
    tooltip=[
        alt.Tooltip('temp_min:Q', title='Min Temperature (°C)', format='.1f'),
        alt.Tooltip('temp_max:Q', title='Max Temperature (°C)', format='.1f'),
        alt.Tooltip('wind:Q', title='Wind Speed', format='.1f'),
        alt.Tooltip('weather:N', title='Weather Type'),
        alt.Tooltip('date:T', title='Date')
    ]
).properties(
    width=180,
    height=140,
    title={
        "text": "Temperature Relationship by Weather Type",
        "subtitle": "Points colored by wind speed with interactive tooltips",
        "fontSize": 16,
        "anchor": "start"
    }
).resolve_scale(
    x='shared',
    y='shared',
    color='shared'
).configure_header(
    titleOrient='bottom',
    titleAnchor='middle',
    titlePadding=10
).configure_axis(
    grid=True,
    gridOpacity=0.3
).configure_view(
    strokeWidth=0
)

chart
```

##### Question 4 (10 points)

Create a compound figure with at least two types of marks. Show the trend of average temperature (the average of min and max temperature) in 4 years. Note, just use the `date` as X attribute, no need to aggragate for years and you will get a wide figure. Include necessary title and labels.

```python
# First, create a new column for average temperature
weather_with_avg = weather.copy()
weather_with_avg['temp_avg'] = (weather_with_avg['temp_min'] + weather_with_avg['temp_max']) / 2

# Create a selection parameter for interactivity
zoom = alt.selection_interval(bind='scales')

# Create the base chart
base = alt.Chart(weather_with_avg)

# Line chart showing the trend
line_chart = base.mark_line(
    color='steelblue',
    strokeWidth=1.5,
    opacity=0.8
).add_params(zoom).encode(
    x=alt.X(
        'date:T',
        title='Date',
        axis=alt.Axis(format='%Y-%m', labelAngle=-45)
    ),
    y=alt.Y(
        'temp_avg:Q',
        title='Average Temperature (°C)',
        scale=alt.Scale(nice=True)
    )
)

# Scatter points to show individual observations
scatter_chart = base.mark_circle(
    size=20,
    opacity=0.6,
    color='orange'
).encode(
    x='date:T',
    y='temp_avg:Q',
    tooltip=[
        alt.Tooltip('date:T', title='Date', format='%Y-%m-%d'),
        alt.Tooltip('temp_avg:Q', title='Average Temperature (°C)', format='.1f'),
        alt.Tooltip('temp_min:Q', title='Min Temperature (°C)', format='.1f'),
        alt.Tooltip('temp_max:Q', title='Max Temperature (°C)', format='.1f'),
        alt.Tooltip('weather:N', title='Weather Type')
    ]
)

# Smooth trend line using LOESS regression
smooth_trend = base.mark_line(
    color='red',
    strokeWidth=3,
    opacity=0.8
).transform_loess(
    'temp_avg', 'date',
    bandwidth=0.1
).encode(
    x='date:T',
    y='temp_avg:Q'
)

# Area chart to show temperature range envelope
area_chart = base.mark_area(
    opacity=0.1,
    color='lightblue'
).encode(
    x='date:T',
    y='temp_min:Q',
    y2='temp_max:Q'
)

# Combine all layers
compound_chart = (area_chart + line_chart + scatter_chart + smooth_trend).resolve_scale(
    color='independent'
).properties(
    width=800,
    height=400,
    title={
        "text": "Average Temperature Trend Over 4 Years",
        "subtitle": "Daily observations (orange dots), daily trend (blue line), and smooth trend (red line)",
        "fontSize": 16,
        "anchor": "start"
    }
).configure_axis(
    grid=True,
    gridOpacity=0.3
).configure_view(
    strokeWidth=0
)

compound_chart
```
